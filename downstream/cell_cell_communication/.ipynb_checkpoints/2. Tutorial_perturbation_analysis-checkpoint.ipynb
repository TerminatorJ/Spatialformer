# Introduction

This jupyter notebook shows a demo of discoverying the effective gene pairs that are important to cell-cell colocalization by running the perturbation analysis of SpatialFormer.

In principle, the perturbation analysis includes the following steps:
1. Replace each gene pairs with their counterparts.
2. Inspecting the perturbation results of the genes.
3. Select the top k most effective gene pairs and analyze their effects.

## Prerequisite:

To analyze the effects of gene pairs, run the following command:

```python
python attention_analysis.py -ct1 T-cells -ct2 Macrophages -cell_num 100
```

WARNING: This script can be executed with any combination of cell types, provided they exist in the slide. By default, the "cell_num" parameter is set to 100. Increasing the "cell_num" will significantly slow down the program but may yield more robust and generalizable results. You can also crop the slide to focus only on specific cells of interest by running the following command:


```python
python attention_analysis.py -ct1 T-cells -ct2 Macrophages -pair_num 100 -cci /scratch/project_465001027/Spatialformer/downstream/cell_cell_communication/data/VUILD96MF_croped_cell_id.npy,
```
In this command, the VUILD96MF_croped_cell_id.npy file contains the fibrosis features in the slides, which exclusively include the cells of Macrophages and T-cells.

After running the script, you will find the result file "all_infos_alllung_crop_id.pkl" in the ./data/ directory.  
The following scripts are based on this generated pickled file.

````python
import pickle
import numpy as np
from tqdm import tqdm
import pandas as pd
from scipy import stats

all_infos = pickle.load(open("/scratch/project_465001027/Spatialformer/downstream/cell_cell_communication/data/all_infos_alllung_crop_id.pkl", "rb"))
````

Loading the token vocabulary

````python
import json
with open("/scratch/project_465001027/Spatialformer/tokenizer/tokenv4.json", "r") as f:
    vocab = json.load(f)
    vocab = {j:i for i,j in vocab.items()}
````

Transfer the results to the dataframe.
- Calculate the mean gene expression level
- Use the student T-test to get the p-value of the gene pairs, and set the null hyphothesis as X~N(0,1)

````python
pair_diff = {}
pair_rank = {}
for cell_pair in all_infos.keys():
    token_pairs = all_infos[cell_pair]["combination_tokens"]
    diffs = all_infos[cell_pair]["diff"]
    rank = all_infos[cell_pair]["combination_indexs"]
    for i, (token_pair, diff) in enumerate(zip(token_pairs, diffs)):
        pairrank = rank[i]
        rev_pair = tuple([token_pair[1], token_pair[0]])
        if rev_pair in pair_diff.keys():
            pair_diff.setdefault(rev_pair,[]).append(diff)
            pair_rank.setdefault(rev_pair,[]).append(pairrank)
        else:
            pair_diff.setdefault(tuple(token_pair),[]).append(diff)
            pair_rank.setdefault(tuple(token_pair),[]).append(pairrank)
#calculating the mean for each pair
pairs = []
mean_diffs = []
mean_ranks = []
support_num = []
t_stats = []
p_values = []
threshold = 0
for pair,diffs in tqdm(pair_diff.items()):
    ranks = pair_rank[pair]
    mean_diff = np.mean(diffs)
    t_stat, p_value = stats.ttest_1samp(diffs, popmean=threshold)
    mean_rank = np.mean(ranks)
#     pair_mean_info.setdefault(pair,{}).
    #transfer the pair to gene symbol
    p_values.append(p_value)
    t_stats.append(t_stat)
    support_num.append(len(diffs))
    gene_pair = (vocab[pair[0]], vocab[pair[1]])
    pairs.append(gene_pair)
    mean_diffs.append(mean_diff)
    mean_ranks.append(mean_rank)

pair_df = pd.DataFrame({"gene_pair": pairs, "Deduction": mean_diffs,"rank_sum": mean_ranks, "stat": t_stats,"P_value": p_values,"support_num": support_num})
````

Finding the overlapping between the pair-wise genes and the ligand-receptor gene pairs

````python
#loading the table of the database
database = pd.read_csv("/scratch/project_465001027/Spatialformer/downstream/cell_cell_communication/data/ligand_receptor_database.csv")
````

````python
# Create a function to check for overlaps
def check_overlap(interaction):
    for i,top_pair in enumerate(pair_df["gene_pair"]):
        if (top_pair[0] in interaction.split("_")) and (top_pair[1] in interaction.split("_")):
            if top_pair[0] != top_pair[1]: #filter
                return int(i)  # Return the top pair if found
    return None  # Return None if no pairs match

# Apply the function to the column to extract overlapping pairs
ovlp_lr_idx = database["interaction.interaction_name"].apply(check_overlap)

# Filter None values to get only the matched pairs
ovlp_lr_idx = np.unique(ovlp_lr_idx.dropna().tolist())  # Convert to a list if needed

# Print matched pairs
print(ovlp_lr_idx)
````

Add the LR info to the dataframe

````python
pair_df["ligand_receptor"] = False
pair_df.loc[ovlp_lr_idx, "ligand_receptor"] = True
#saing the pair dataframe
pair_df.to_csv("/scratch/project_465001027/Spatialformer/downstream/cell_cell_communication/data/perturbation_pairs.csv")
````

Transform the "Deduction" column into "10log(1-D)" and keep sign

````python
pair_df['10log(1-D)'] = -10*np.log(1-np.abs(pair_df['Deduction'])) / np.log(2)  # Apply -log transformation
pair_df['symbol'] = [1 if i >0 else -1 for i in pair_df['Deduction']]
pair_df['10log(1-D)'] = pair_df['10log(1-D)']*pair_df['symbol']
````

````python
%%R -w 500 -h 500
library(ggplot2)
# library(ggrepel)  # For better text label placement
library(dplyr)    # For data manipulation
library(MASS)     # For kde2d
library(ggpointdensity)
library(viridis)
pair_df$Transformed_Deduction <- -log(1-(-pair_df$Deduction)) #only select minus pairs
# horizon_line <- 0
# Perform linear regression
model <- lm(Transformed_Deduction ~ rank_sum, data = pair_df)
summary_model <- summary(model)
r_squared <- summary_model$r.squared
p_value <- summary_model$coefficients[2, 4]  # p-value for the rank_sum coefficient\
print(p_value)
# # Identify the top 2 pairs with max log(Deduction)
top_pairs <- pair_df[order(pair_df$Transformed_Deduction, decreasing = TRUE), ][1:2, ]
print(top_pairs)
# Add a column for color based on p-value
# Add a column for color based on p-value and your conditions
pair_df$Color <- NA
pair_df$Color[pair_df$P_value < 0.05 & pair_df$Deduction < 0 & pair_df$stat < 0 & pair_df$support_num > 50] <- "red"
# Map colors: significant points are in red with a gradient based on significance
pair_df$Color <- ifelse(is.na(pair_df$Color), "grey", pair_df$Color)  # Non-significant = grey
# head(df)
# Set non-significant points to grey
pair_df$Color[is.na(pair_df$Color)] <- "grey"
num_significant <- sum(pair_df$Color == "red")
# df$Significance_Score <- 1 / df$P_value
# Create the scatter plot
p <- ggplot(pair_df, aes(x = rank_sum, y = Transformed_Deduction)) +
    geom_point(data = pair_df[pair_df$Color == "grey", ], aes(color = "grey"), size = 1, alpha = 0.6) +  # Plot grey dots first
    geom_point(data = pair_df[pair_df$Color == "red", ], aes(color = "red"), size = 1, alpha = 0.8) +  # Overlay red dots
    scale_color_manual(values = c("grey" = "grey", "red" = "red")) +  # Define explicit colors
    geom_smooth(method = "lm", color = "black", se = FALSE) +  # Add linear regression line
    geom_hline(yintercept = 0, linetype = "dashed", color = "black") +  # Horizontal line at y=0
     labs(x = "Rank Sum", 
         y = "-log(1-Deduction)", 
         title = paste("Macrophages vs T-cells", "\nR² =", round(r_squared, 3), " p-value =", formatC(p_value, format = "e", digits = 3), "n =", num_significant)) +
    theme_minimal() +
    theme(
        plot.margin = margin(t = 50, r = 20, b = 20, l = 20),  # Increase margin
        axis.title.x = element_text(size = 14),    # Increase x-axis label size
        axis.title.y = element_text(size = 14),    # Increase y-axis label size
        axis.text.x = element_text(size = 12),      # Increase x-axis tick label size
        axis.text.y = element_text(size = 12),
        axis.line = element_line(colour = "black"),
        panel.border = element_blank(),  # Remove panel border
        panel.grid = element_blank(),     # Remove grid lines
        axis.ticks.length = unit(0.25, "cm")   # Adjust tick length if needed
    )
ggsave("/scratch/project_465001027/Spatialformer/downstream/cell_cell_communication/figures/macrophages_vs_tcells_plot.png", plot = p, width = 8, height = 8, dpi = 300, bg = "white") 
print(p)
````
