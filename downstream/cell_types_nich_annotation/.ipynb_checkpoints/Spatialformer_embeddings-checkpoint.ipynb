```python
def get_dataset(X):
    """
    Getting the dataset for the leave -ut sample - "VUILD110"
    This dataset will be used to calculate the embeddings and the downstream cell types/niches annotations

    Parameters:
    X: The expression matrix with cell x gene matrix that comes from the leave-out sample/slide.

    """
    sample_name = "VUILD110"
    cell_ids = list(X.index)
    #loading the dataset
    datapath = "/scratch/project_465001820/Spatialformer/cache/" # Customise your own path here to store the cache of the dataset, otherwise, it will be saved to the root dir.
    combined_dataset = load_dataset("TerminatorJ/xenium_pandavid_dataset4", cache_dir = datapath, num_proc=8)
    index_path = "/scratch/project_465001820/Spatialformer/data/sample_cell_index.pkl"
    sample_cell_index = get_index(combined_dataset, save_file = index_path)
    index = [sample_cell_index[sample_name][cell_id] for cell_id in cell_ids]

    combined_dataset_all = concatenate_datasets([combined_dataset["train"], combined_dataset["test"], combined_dataset["validation"]])
    dataset = combined_dataset_all.select(index)
    return dataset

```

Loading the model 
- The checkpoint of the singular input model can be downloaded via: https://figshare.com/articles/dataset/single_input/28452209?file=52503695)

```python
def load_model(model_ckp_path, device):
    get_file_path = lambda path, filename: os.path.join("/scratch/project_465001820/Spatialformer", path, filename)
    config_path = get_file_path("config", "_config_train_large_pair.json")
    with open(config_path, 'r') as json_file:
        config = json.load(json_file)
    model = manual_train_fm(config = config)
    ckp = torch.load(model_ckp_path, map_location=torch.device(device))
    params = ckp["state_dict"]
    model.load_state_dict(params)
    model.eval()
    model.to(device)
    return model

#loading the single model
model_ckp_path = "/scratch/project_465001820/Spatialformer/output/checkpoints/step=0104000-train_total_loss=-2.3064-val_total_loss=0.0000.ckpt"
model = load_model(model_ckp_path, "cuda")
```

# Getting the cell type embeddings for SpatialFormer

```python
%%time
#get train test val dataset
train_ct_dataset = get_dataset(ct_train_X)
test_ct_dataset = get_dataset(ct_test_X)
val_ct_dataset = get_dataset(ct_val_X)

batch_size = 8
train_ct_dataloader = create_single_data_loaders(train_ct_dataset, cls_token = 1, padding_idx = 0, 
                           sep_token = 1949, batch_size=batch_size, context_length=500, 
                           special_token_num = 4, split_num = 1, num_workers = 16)



test_ct_dataloader = create_single_data_loaders(test_ct_dataset, cls_token = 1, padding_idx = 0, 
                           sep_token = 1949, batch_size=batch_size, context_length=500, 
                           special_token_num = 4, split_num = 1, num_workers = 16)


val_ct_dataloader = create_single_data_loaders(val_ct_dataset, cls_token = 1, padding_idx = 0, 
                           sep_token = 1949, batch_size=batch_size, context_length=500, 
                           special_token_num = 4, split_num = 1, num_workers = 16)



train_ct_embeds1 = []
train_ct_embeds2 = []
train_ct_embeds3 = []
train_ct_anns = []
with torch.no_grad(): 
    for i, batch in tqdm(enumerate(train_ct_dataloader), total = int(train_ct_dataset.shape[0]/batch_size)):
        attention_mask = batch["attention_mask"]
        attention_mask1 = batch["attention_mask"]
        attention_mask2 = batch["attention_mask"]
        
        #getting the meaningful mask
        #release the mask only for meaningful tokens
        for i,mask in enumerate(attention_mask1):
            sequence_len = attention_mask[i].sum()
            attention_mask1[i][:1] = False #only cls
            attention_mask1[i][sequence_len-1] = False #mute the sep token
            attention_mask2[i][:5] = False #mute the cls and meta
            attention_mask2[i][sequence_len-1] = False #mute the sep token
        anns = batch["Annotations"]
        attention_mask_sq1 = attention_mask1.unsqueeze(-1).cuda()  # Shape: [80, 400, 1]
        attention_mask_sq2 = attention_mask2.unsqueeze(-1).cuda()  # Shape: [80, 400, 1]
        embeddings, ref_pair_prob = model.get_embeddings(batch, [-1], True, False) #normal prob
        #getting the conditions and tissue types
        embeddings1 = embeddings[0][:,0,:]
        embeddings1 = embeddings1.detach().cpu().numpy()#The embeddings should 
    
        train_ct_embeds1.append(embeddings1)
        #getting the gene tokens embeddings
        embeddings2 = torch.sum(embeddings[0] * attention_mask_sq1, dim=1) / torch.sum(attention_mask_sq1, dim=1)
        embeddings2 = embeddings2.detach().cpu().numpy()
        train_ct_embeds2.append(embeddings2)
        
        embeddings3 = torch.sum(embeddings[0] * attention_mask_sq2, dim=1) / torch.sum(attention_mask_sq2, dim=1)
        embeddings3 = embeddings3.detach().cpu().numpy()
        train_ct_embeds3.append(embeddings3)
        
        train_ct_anns.append(anns)


test_ct_embeds1 = []
test_ct_embeds2 = []
test_ct_embeds3 = []
test_ct_embeds4 = []
test_ct_embeds5 = []
test_ct_anns = []
with torch.no_grad(): 
    for i, batch in tqdm(enumerate(test_ct_dataloader), total = int(test_ct_dataset.shape[0]/batch_size)):
        attention_mask = batch["attention_mask"]
        attention_mask1 = batch["attention_mask"]
        attention_mask2 = batch["attention_mask"]

        
        #getting the meaningful mask
        #release the mask only for meaningful tokens
        for i,mask in enumerate(attention_mask1):
            sequence_len = attention_mask[i].sum()
            attention_mask1[i][:1] = False #only mute cls
            attention_mask1[i][sequence_len-1] = False #mute the sep token
            attention_mask2[i][:5] = False #mute the cls and meta
            attention_mask2[i][sequence_len-1] = False #mute the sep token
        anns = batch["Annotations"]
        attention_mask_sq1 = attention_mask1.unsqueeze(-1).cuda()  # Shape: [80, 400, 1]
        attention_mask_sq2 = attention_mask2.unsqueeze(-1).cuda()  # Shape: [80, 400, 1]
        embeddings, ref_pair_prob = model.get_embeddings(batch, [-1], True, False) #normal prob
        all_embeddings, ref_pair_prob = model.get_embeddings(batch, [i for i in range(10)], True, False) #normal prob
        #getting the conditions and tissue types
        embeddings1 = embeddings[0][:,0,:]
        embeddings1 = embeddings1.detach().cpu().numpy()#The embeddings should 
        
    
        test_ct_embeds1.append(embeddings1)
        #getting the gene tokens embeddings
        embeddings2 = torch.sum(embeddings[0] * attention_mask_sq1, dim=1) / torch.sum(attention_mask_sq1, dim=1)
        embeddings2 = embeddings2.detach().cpu().numpy()
        test_ct_embeds2.append(embeddings2)
        
        embeddings3 = torch.sum(embeddings[0] * attention_mask_sq2, dim=1) / torch.sum(attention_mask_sq2, dim=1)
        embeddings3 = embeddings3.detach().cpu().numpy()
        test_ct_embeds3.append(embeddings3)
        
        embeddings4 = torch.mean(embeddings[0][:,5:,:], dim=1)
        embeddings4 = embeddings4.detach().cpu().numpy()
        test_ct_embeds4.append(embeddings4)
        
        combined_tensor = torch.stack(all_embeddings, dim = -1)
        embeddings5 = torch.mean(combined_tensor, dim = -1)[:,0,:]  # bs x seq_len x dim
        embeddings5 = embeddings5.detach().cpu().numpy()
        test_ct_embeds5.append(embeddings5)
        
        test_ct_anns.append(anns)
        
val_ct_embeds1 = []
val_ct_embeds2 = []
val_ct_embeds3 = []
val_ct_anns = []
with torch.no_grad(): 
    for i, batch in tqdm(enumerate(val_ct_dataloader), total = int(val_ct_dataset.shape[0]/batch_size)):
        attention_mask = batch["attention_mask"]
        attention_mask1 = batch["attention_mask"]
        attention_mask2 = batch["attention_mask"]
        
        #getting the meaningful mask
        #release the mask only for meaningful tokens
        for i,mask in enumerate(attention_mask1):
            sequence_len = attention_mask[i].sum()
            attention_mask1[i][:1] = False #only cls
            attention_mask1[i][sequence_len-1] = False #mute the sep token
            attention_mask2[i][:5] = False #mute the cls and meta
            attention_mask2[i][sequence_len-1] = False #mute the sep token
        anns = batch["Annotations"]
        attention_mask_sq1 = attention_mask1.unsqueeze(-1).cuda()  # Shape: [80, 400, 1]
        attention_mask_sq2 = attention_mask2.unsqueeze(-1).cuda()  # Shape: [80, 400, 1]
        embeddings, ref_pair_prob = model.get_embeddings(batch, [-1], True, False) #normal prob
        #getting the conditions and tissue types
        embeddings1 = embeddings[0][:,0,:]
        embeddings1 = embeddings1.detach().cpu().numpy()#The embeddings should 
    
        val_ct_embeds1.append(embeddings1)
        #getting the gene tokens embeddings
        embeddings2 = torch.sum(embeddings[0] * attention_mask_sq1, dim=1) / torch.sum(attention_mask_sq1, dim=1)
        embeddings2 = embeddings2.detach().cpu().numpy()
        val_ct_embeds2.append(embeddings2)
        
        embeddings3 = torch.sum(embeddings[0] * attention_mask_sq2, dim=1) / torch.sum(attention_mask_sq2, dim=1)
        embeddings3 = embeddings3.detach().cpu().numpy()
        val_ct_embeds3.append(embeddings3)
        
        val_ct_anns.append(anns)
```

Concate results


```python
import numpy as np
train_ct_concat_embed1 = np.vstack(train_ct_embeds1)
train_ct_concat_embed2 = np.vstack(train_ct_embeds2)
train_ct = [a_ for a in train_ct_anns for a_ in a]

test_ct_concat_embed1 = np.vstack(test_ct_embeds1)
test_ct_concat_embed2 = np.vstack(test_ct_embeds2)
test_ct_concat_embed3 = np.vstack(test_ct_embeds3)
test_ct_concat_embed4 = np.vstack(test_ct_embeds4)
test_ct_concat_embed5 = np.vstack(test_ct_embeds5)
test_ct = [a_ for a in test_ct_anns for a_ in a]

val_ct_concat_embed1 = np.vstack(val_ct_embeds1)
val_ct_concat_embed2 = np.vstack(val_ct_embeds2)
val_ct = [a_ for a in val_ct_anns for a_ in a]
```

getting the cell niches embeddings

```python
%%time
#get train test val dataset
train_nc_dataset = get_dataset(nc_train_X)
test_nc_dataset = get_dataset(nc_test_X)
val_nc_dataset = get_dataset(nc_val_X)

#building the dataloader
batch_size = 8
train_nc_dataloader = create_single_data_loaders(train_nc_dataset, cls_token = 1, padding_idx = 0, 
                           sep_token = 1949, batch_size=batch_size, context_length=500, 
                           special_token_num = 4, split_num = 1, num_workers = 16)


test_nc_dataloader = create_single_data_loaders(test_nc_dataset, cls_token = 1, padding_idx = 0, 
                           sep_token = 1949, batch_size=batch_size, context_length=500, 
                           special_token_num = 4, split_num = 1, num_workers = 16)


val_nc_dataloader = create_single_data_loaders(val_nc_dataset, cls_token = 1, padding_idx = 0, 
                           sep_token = 1949, batch_size=batch_size, context_length=500, 
                           special_token_num = 4, split_num = 1, num_workers = 16)


# %%time
train_nc_embeds1 = []
train_nc_embeds2 = []
train_nc_anns = []
with torch.no_grad(): 
    for i, batch in tqdm(enumerate(train_nc_dataloader), total = int(train_nc_dataset.shape[0]/batch_size)):
        attention_mask = batch["attention_mask"]

        #getting the meaningful mask
        #release the mask only for meaningful tokens
        for i,mask in enumerate(attention_mask):
            sequence_len = attention_mask[i].sum()
            attention_mask[i][:5] = False #mute the cls and meta
            attention_mask[i][sequence_len-1] = False #mute the sep token

        anns = batch["Niche_Annotations"]
        attention_mask_sq = attention_mask.unsqueeze(-1).cuda()  # Shape: [8, 500, 512]
        embeddings, ref_pair_prob = model.get_embeddings(batch, [-1], True, False) #normal prob
        #getting the conditions and tissue types
        #getting the cls token embeddings
        embeddings1 = embeddings[0][:,0,:]
        embeddings1 = embeddings1.detach().cpu().numpy()#The embeddings should 

        train_nc_embeds1.append(embeddings1)
        #getting the gene tokens embeddings
        embeddings2 = torch.sum(embeddings[0] * attention_mask_sq, dim=1) / torch.sum(attention_mask_sq, dim=1)
        embeddings2 = embeddings2.detach().cpu().numpy()
        train_nc_embeds2.append(embeddings2)

        train_nc_anns.append(anns)


test_nc_embeds1 = []
test_nc_embeds2 = []
test_nc_anns = []
with torch.no_grad(): 
    for i, batch in tqdm(enumerate(test_nc_dataloader), total = int(test_nc_dataset.shape[0]/batch_size)):
        attention_mask = batch["attention_mask"]
        
        #getting the meaningful mask
        #release the mask only for meaningful tokens
        for i,mask in enumerate(attention_mask):
            sequence_len = attention_mask[i].sum()
            attention_mask[i][:5] = False #mute the cls and meta
            attention_mask[i][sequence_len-1] = False #mute the sep token
        anns = batch["Niche_Annotations"]
        attention_mask_sq = attention_mask.unsqueeze(-1).cuda()  # Shape: [80, 400, 1]
        embeddings, ref_pair_prob = model.get_embeddings(batch, [-1], True, False) #normal prob
        #getting the conditions and tissue types
        embeddings1 = embeddings[0][:,0,:]
        embeddings1 = embeddings1.detach().cpu().numpy()#The embeddings should 
    
        test_nc_embeds1.append(embeddings1)
        #getting the gene tokens embeddings
        embeddings2 = torch.sum(embeddings[0] * attention_mask_sq, dim=1) / torch.sum(attention_mask_sq, dim=1)
        embeddings2 = embeddings2.detach().cpu().numpy()
        test_nc_embeds2.append(embeddings2)
        
        test_nc_anns.append(anns)
        
val_nc_embeds1 = []
val_nc_embeds2 = []
val_nc_anns = []
with torch.no_grad(): 
    for i, batch in tqdm(enumerate(val_nc_dataloader), total = int(val_nc_dataset.shape[0]/batch_size)):
        attention_mask = batch["attention_mask"]
        for i,mask in enumerate(attention_mask):
            sequence_len = attention_mask[i].sum()
            attention_mask[i][:5] = False #mute the cls and meta
            attention_mask[i][sequence_len-1] = False #mute the sep token
        attention_mask_sq = attention_mask.unsqueeze(-1).cuda()  # Shape: [80, 400, 1]
        anns = batch["Niche_Annotations"]
        embeddings, ref_pair_prob = model.get_embeddings(batch, [-1], True, False) #normal prob
        #getting the conditions and tissue types
        embeddings1 = embeddings[0][:,0,:]
        embeddings1 = embeddings1.detach().cpu().numpy()#The embeddings should 
    
        val_nc_embeds1.append(embeddings1)
        embeddings2 = torch.sum(embeddings[0] * attention_mask_sq, dim=1) / torch.sum(attention_mask_sq, dim=1)
        embeddings2 = embeddings2.detach().cpu().numpy()
        val_nc_embeds2.append(embeddings2)
        
        val_nc_anns.append(anns)
```

Concate niche embeddings results

```python
import numpy as np
train_nc_concat_embed1 = np.vstack(train_nc_embeds1)
train_nc_concat_embed2 = np.vstack(train_nc_embeds2)
train_nc = [a_ for a in train_nc_anns for a_ in a]

test_nc_concat_embed1 = np.vstack(test_nc_embeds1)
test_nc_concat_embed2 = np.vstack(test_nc_embeds2)
test_nc = [a_ for a in test_nc_anns for a_ in a]

val_nc_concat_embed1 = np.vstack(val_nc_embeds1)
val_nc_concat_embed2 = np.vstack(val_nc_embeds2)
val_nc = [a_ for a in val_nc_anns for a_ in a]
```
