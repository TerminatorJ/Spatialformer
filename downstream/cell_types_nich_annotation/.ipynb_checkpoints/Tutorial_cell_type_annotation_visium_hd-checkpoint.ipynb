# Visium HD dataset preprocessing

All the Visium HD data of Colorectal cancer(CRC) are downloaded from GEO under accession number GSE280318  
The publication can be accessed DOI: https://doi.org/10.1101/2024.06.04.597233

```python
import scanpy as sc
import pandas as pd
import numpy as np
import anndata as ad
```

```python
adata_p1 = sc.read_10x_h5("/scratch/project_465001820/Spatialformer/data/Visium_HD/GSE280318/GSM8594567_P1CRC_filtered_feature_bc_matrix.h5")
meta_p1 = pd.read_parquet("/scratch/project_465001820/Spatialformer/data/Visium_HD/GSE280318/GSM8594567_P1CRC_Metadata.parquet")

adata_p2 = sc.read_10x_h5("/scratch/project_465001820/Spatialformer/data/Visium_HD/GSE280318/GSM8594568_P2CRC_filtered_feature_bc_matrix.h5")
meta_p2 = pd.read_parquet("/scratch/project_465001820/Spatialformer/data/Visium_HD/GSE280318/GSM8594568_P2CRC_Metadata.parquet")

adata_p3 = sc.read_10x_h5("/scratch/project_465001820/Spatialformer/data/Visium_HD/GSE280318/GSM8594569_P5CRC_filtered_feature_bc_matrix.h5")
meta_p3 = pd.read_parquet("/scratch/project_465001820/Spatialformer/data/Visium_HD/GSE280318/GSM8594569_P5CRC_Metadata.parquet")
```

Merge metadata with the expression matrix

```python
meta_p1_aligned = meta_p1.set_index("barcode").loc[adata_p1.obs.index.astype(str)]
# Assign it to adata_p1.obs
adata_p1.obs = meta_p1_aligned

meta_p2_aligned = meta_p2.set_index("barcode").loc[adata_p2.obs.index.astype(str)]
# Assign it to adata_p1.obs
adata_p2.obs = meta_p2_aligned

meta_p3_aligned = meta_p3.set_index("barcode").loc[adata_p3.obs.index.astype(str)]
# Assign it to adata_p1.obs
adata_p3.obs = meta_p3_aligned
```

```python
# Ensure unique obs_names in each AnnData object by prefixing them
adata_p1.obs_names = [f"sample1_{x}" for x in adata_p1.obs_names]
adata_p2.obs_names = [f"sample2_{x}" for x in adata_p2.obs_names]
adata_p3.obs_names = [f"sample3_{x}" for x in adata_p3.obs_names]
```

Filter out the duplicated genes

```python
adata_p1 = adata_p1[:, ~adata_p1.var.index.duplicated()]
adata_p2 = adata_p2[:, ~adata_p2.var.index.duplicated()]
adata_p3 = adata_p3[:, ~adata_p3.var.index.duplicated()]
```

```python
import scanpy as sc
# Concatenate the AnnData objects
adata_combined = sc.concat(
    [adata_p1, adata_p2, adata_p3],
    axis=0,                 # Concatenate observations
    join='inner',           # Keeps intersection of genes/features
    label='sample',         # Adds a column in .obs to distinguish samples
    keys=['sample1', 'sample2', 'sample3']  # Sample labels
)

# Verify structure
print(adata_combined)
```

```python
adata_combined.obs["MacrophageSubtype"].unique()
```

Assign macrophage subtypes to the macrophage clusters

```python
adata_combined.obs["DeconvolutionLabel1"][adata_combined.obs["MacrophageSubtype"] == "Macrophage-SPP1+"] = "Macrophage-SPP1+"
adata_combined.obs["DeconvolutionLabel1"][adata_combined.obs["MacrophageSubtype"] == "Macrophage-SELENOP+"] = "Macrophage-SELENOP+"
```

```python
adata_combined.obs["DeconvolutionLabel1"].unique()
```

```python
adata_combined.obs
```

Subset 10% of the data according to the cell types (deconvolution)

```python
adata_combined
```

```python
# Group cells by the labels in "DeconvolutionLabel1"
groups = adata_combined.obs.groupby("DeconvolutionLabel1")
np.random.seed(42)
# Subset 5% of cells from each group
subset_indices = []
for label, group in groups:
    # Randomly select 5% of cells from the group
    subset_indices.extend(
        np.random.choice(group.index, size=int(len(group) * 0.05), replace=False)
    )

# Create a subset of the AnnData object
adata_subset = adata_combined[subset_indices].copy()

# Verify the subset structure
print(adata_subset)
```

Filter out the cell types with the cells less than 20

```python
cts = adata_subset.obs["DeconvolutionLabel1"].value_counts()[adata_subset.obs["DeconvolutionLabel1"].value_counts() < 20]
```

```python
adata_subset_flt = adata_subset[~adata_subset.obs["DeconvolutionLabel1"].isin(cts.index),:]
```

```python
adata_subset_flt
```

Split the data into train-test-validation

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

# Define proportions
train_ratio = 0.7  # 70% for training
test_ratio = 0.2   # 20% for testing
val_ratio = 0.1    # 10% for validation

# Extract labels for stratified splitting
labels = adata_subset_flt.obs["DeconvolutionLabel1"]

# First split: Train and Temp (Test + Validation)
train_indices, temp_indices = train_test_split(
    adata_subset_flt.obs.index,
    test_size=(test_ratio + val_ratio),  # Remaining 30%
    stratify=labels,
    random_state=42  # Ensure reproducibility
)

# Second split: Test and Validation
test_indices, val_indices = train_test_split(
    temp_indices,
    test_size=(val_ratio / (test_ratio + val_ratio)),  # Adjust proportion for validation
    stratify=labels[temp_indices],
    random_state=42  # Ensure reproducibility
)

# Create subsets
adata_train = adata_subset_flt[train_indices].copy()
adata_test = adata_subset_flt[test_indices].copy()
adata_val = adata_subset_flt[val_indices].copy()

# Verify structure of subsets
print(adata_train)
print(adata_test)
print(adata_val)
```

Extract the matrix

```python
train_X = pd.DataFrame(adata_train.X.todense(), columns=adata_train.var.index, index=adata_train.obs.index)
```

```python
test_X = pd.DataFrame(adata_test.X.todense(), columns=adata_test.var.index, index=adata_test.obs.index)
```

```python
val_X = pd.DataFrame(adata_val.X.todense(), columns=adata_val.var.index, index=adata_val.obs.index)
```

```python
def reassign_mac_color(adata_ct_mac, pred_column):
    pred_mac_cts = adata_ct_mac.obs[pred_column].cat.categories
    pred_mac_colors = adata_ct_mac.uns[f"{pred_column}_colors"]
    defined_colors = ['#AEEEEE', '#32CD32']
    mac_cts = ["Macrophage", "Macrophage-SPP1+"]
    mac_dict = dict(zip(mac_cts, defined_colors))
    mac_idx = [i for i, ct in enumerate(pred_mac_cts) if ct in mac_cts]
    mac_colors = [mac_dict[ct] for i, ct in enumerate(pred_mac_cts) if ct in mac_cts]
#     mac_ct = [i for i, ct in enumerate(pred_mac_cts) if ct in mac_cts]
    other_idx = [i for i, ct in enumerate(pred_mac_cts) if ct not in mac_cts]
    # new_pred_mac_colors = [i for i, c in enumerate(pred_mac_colors) if i in mac_idx]
    pred_mac_colors[mac_idx] = mac_colors
    pred_mac_colors[other_idx] = "#BEBEBE"
    adata_ct_mac.uns[f"{pred_column}_colors"] = pred_mac_colors
    return adata_ct_mac
```
