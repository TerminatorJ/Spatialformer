```python
#loading the embeddings of Spatialformer
import numpy as np
import scanpy as sc
import matplotlib.pyplot as plt
test_spa_embed = np.load("/home/sxr280/Spatialformer/data/test_embedding_True_False_8_0.1_2_None_20240908_091236.npy")
test_spa_ann = np.load('/home/sxr280/Spatialformer/data/test_labels_True_False_8_0.1_2_None_20240908_091236.npy')
adata_spa_raw = sc.AnnData(test_spa_embed)
adata_spa = adata_spa_raw.copy()
```

```python
test_spa_embed.shape
```

```python
#plotting the cell type annotations
sc.pp.neighbors(adata_spa,use_rep='X')

sc.tl.umap(adata_spa)

adata_spa.obs['true_label']=test_spa_ann
# zadata.obs['pred_label']=zheng_test_str[y_pred]
# adata = adata[adata.obs['true_label']=="Fibroblasts"]

plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['ps.fonttype'] = 42
sc.pl.umap(adata_spa,color=['true_label'],wspace=0.3,size=50,save='scf_seg')
```

```python
test_scGPT_embed = np.load("/home/sxr280/Spatialformer/data/test_embeddings_scGPT.npy")
test_scGPT_ann = np.load('/home/sxr280/Spatialformer/data/test_frac10_ann.npy')
adata_scGPT_raw = sc.AnnData(test_scGPT_embed)
adata_scGPT = adata_scGPT_raw.copy()
#plotting the cell type annotations
sc.pp.neighbors(adata_scGPT,use_rep='X')

sc.tl.umap(adata_scGPT)

adata_scGPT.obs['true_label']=test_scGPT_ann

plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['ps.fonttype'] = 42
sc.pl.umap(adata_scGPT,color=['true_label'],wspace=0.3,size=50,save='scf_seg')
```

```python
# Encode labels
from sklearn.preprocessing import LabelEncoder
scGPT_pred = np.load("/home/sxr280/Spatialformer/downstream/cell_types_annotation/data/all_preds_1_scGPT_20240908_230116.npy")
label_encoder = LabelEncoder()
test_labels_encoded = label_encoder.fit_transform(test_scGPT_ann)
unique_label_names = label_encoder.classes_
from sklearn.metrics import classification_report
print(classification_report(test_labels_encoded, scGPT_pred, target_names=unique_label_names))
scGPT_pred_ann = [unique_label_names[i] for i in scGPT_pred]
adata_scGPT.obs['pred_label']=scGPT_pred_ann
adata_scGPT = adata_scGPT[adata_scGPT.obs['true_label']=="Fibroblasts"]
plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['ps.fonttype'] = 42
sc.pl.umap(adata_scGPT,color=['pred_label'],wspace=0.3,size=50,save='scf_seg')

```
