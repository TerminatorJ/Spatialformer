# Getting the expression matrix

```r
expression_data <- GetAssayData(xenium[["cell_RNA"]], slot = "count")
# expression_data_df <- as.data.frame(expression_data)
sorted_indices <- order(rownames(expression_data))
expression_data_sorted <- expression_data[sorted_indices, ]
# expression_data_sorted <- expression_data1[order(rownames(expression_data1)), ]
```

```r
library(Matrix)
# Save the sparse matrix in Matrix Market format
writeMM(expression_data_sorted, file = "/home/sxr280/Spatialformer/data/expression_data.mtx")
```

# Getting the metadata

```r

# xenium@meta.data$cell_id ,xenium@meta.data$broad_CT5
nich_annotation <- data.frame(
  cell_id = row.names(xenium@meta.data),
  broad_CT5 = xenium@meta.data$TNiche
)
write.csv(nich_annotation, "/home/sxr280/Spatialformer/data/nich_annotation_df.csv", row.names=FALSE)
```

```r

# xenium@meta.data$cell_id ,xenium@meta.data$broad_CT5
annotation <- data.frame(
  cell_id = row.names(xenium@meta.data),
  broad_CT5 = xenium@meta.data$broad_CT5
)
write.csv(annotation, "/home/sxr280/Spatialformer/data/annotation_df.csv", row.names=FALSE)
```
