```python
import umap
import umap.plot
import pickle
from datasets import load_from_disk
import numpy as np
import matplotlib.pyplot as plt
```

```python
#loading the selected dataset, which has all the info we need
combined_dataset = load_from_disk("/home/sxr280/Spatialformer/data/embedding_visualization_arrow")
embeddings_list = combined_dataset["Embeddings_Norm"]

# Convert list of lists to a NumPy array with shape (num_samples, 100)
embeddings_array = np.array(embeddings_list)
#Getting the annotations
annotation = np.array(combined_dataset["Annotations"])
sample_names = combined_dataset["Sample_Names"]
conditions = ["Healthy" if "HD" in i else "Disease" for i in sample_names]
```

```python
import umap
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from sklearn.preprocessing import LabelEncoder
import random

# Assuming `embeddings_array` and `annotation` are defined
# Fit UMAP
n = 8  # Example value for n_neighbors
m = 0.3  # Example value for min_dist

mapper = umap.UMAP(n_neighbors=n, min_dist=m).fit(embeddings_array)

# Extract UMAP embedding
embedding = mapper.transform(embeddings_array)

# Encode the labels
label_encoder = LabelEncoder()
encoded_labels = label_encoder.fit_transform(annotation)
unique_labels = label_encoder.classes_
# Define custom colors
colors = ['#EFA7AA', '#74AB6D', '#643A90', '#2575B2', '#F08E40', '#D62F81', '#349979', '#8B95F2', '#6EA0A0', '#CC7276',
          '#F5BA6E', '#F07858', '#C3DDA5', '#64BA9F', '#DDA713', '#ED8760', '#A8DDB5', '#B5382C', '#C6C6C5', '#E2DCF0',
          '#EED38B', '#B1CF98', '#EA97C0', '#86B6D3', '#489B90', '#9E9C3D', '#CECD9E', '#516B84', '#E3DEF5', '#DBE98A',
          '#70C26A', '#DCC0DC', '#F1C3BD', '#F5EFB4', '#0AB0C8', '#3C6E80', '#7F9EAC', '#AAB3B2', '#8BC63E']  # Add more colors if needed
# colors = ['#EFA7AA', '#74AB6D']

# Define the number of colors needed
n_colors = len(np.unique(encoded_labels))

## Create a colormap
# cmap = plt.get_cmap('Set1')
cmap1 = plt.get_cmap('Pastel2')
cmap2 = plt.get_cmap('Paired')
cmap3 = plt.get_cmap('Accent')
cmap4 = plt.get_cmap('Set3')
# Generate random indices to sample colors
# random_indices = np.random.rand(n_colors)
# colors = [cmap1(i/n_colors) for i in range(n_colors)] + [cmap2(i/n_colors) for i in range(n_colors)] + [cmap3(i/n_colors) for i in range(n_colors)]
# colors = np.unique(colors)[:n_colors]

# Create a mapping from encoded labels to colors
label_to_color = dict(zip(np.unique(encoded_labels), colors))

# Apply the color mapping to the encoded labels for the scatter plot
color_for_scatter = [label_to_color[label] for label in encoded_labels]

# Plot
plt.figure(figsize=(13, 10))
scatter = plt.scatter(embedding[:, 0], embedding[:, 1], c=color_for_scatter, s=1)

# Create legend handles with custom colors

handles = [plt.Line2D([0], [0], color=label_to_color[label], linestyle='', markersize=6, marker='o') for label in range(n_colors)]

# Add the legend
plt.legend(handles, unique_labels, title="Cell Types", bbox_to_anchor=(1, 1), loc='upper left')

plt.tight_layout()
plt.savefig("/home/sxr280/Spatialformer/Figures/embeddings_visualization_cell_types_10_0.5.png")
plt.show()
```

```python

import time
import matplotlib.pyplot as plt
import umap
import umap.plot
import matplotlib.pyplot as plt
import networkx as nx
import community.community_louvain as community_louvain
 

normal_label_dict = {}
simulation_label_dict = {}
for i,ann in enumerate(np.unique(annotation)[:10]):
    #filter the clusters
    print(f"doing {ann}")
    select_idx = np.where(np.array(annotation_nb) == ann)[0]
    embeddings_s = embeddings_array_nb[select_idx]
    mapper = umap.UMAP(n_neighbors = 10, min_dist = 0.5).fit(embeddings_s)
    knn_graph = umap.umap_.fuzzy_simplicial_set(
    X=embeddings_s,
    n_neighbors=100,
    random_state=np.random.RandomState(),
    metric='euclidean')[0]

    # Convert knn_graph to a networkx graph
    G = nx.Graph(knn_graph)

    # Step 4: Apply Louvain clustering
    louvain_partition = community_louvain.best_partition(G)

    # Extract the cluster labels
    cluster_labels = np.array(list(louvain_partition.values()))

    # Step 5: Plot the UMAP embedding with Louvain clusters
    fig1, ax1 = plt.subplots()
    print("normal")
    normal_label_dict[ann] = len(np.unique(cluster_labels))
    fig1 = umap.plot.points(mapper, labels=cluster_labels, color_key_cmap='Set1')
    plt.show()
#####################################################################################
    select_idx = np.where(np.array(annotation_nbp) == ann)[0]
    embeddings_s = embeddings_array_nbp[select_idx]
    mapper = umap.UMAP(n_neighbors = 10, min_dist = 0.5).fit(embeddings_s)
    knn_graph = umap.umap_.fuzzy_simplicial_set(
    X=embeddings_s,
    n_neighbors=100,
    random_state=np.random.RandomState(),
    metric='euclidean')[0]

    # Convert knn_graph to a networkx graph
    G = nx.Graph(knn_graph)

    # Step 4: Apply Louvain clustering
    louvain_partition = community_louvain.best_partition(G)

    # Extract the cluster labels
    cluster_labels = np.array(list(louvain_partition.values()))

    # Step 5: Plot the UMAP embedding with Louvain clusters
    fig2, ax2 = plt.subplots()
    print("simulation")
    simulation_label_dict[ann] = len(np.unique(cluster_labels))
    fig2 = umap.plot.points(mapper, labels=cluster_labels, color_key_cmap='Set1')
    plt.show()
    plt.close()
    # time.sleep(3)


```

```python
def get_embedding(embeddings, annotation, cluster):
    select_idx = np.where(np.array(annotation) == cluster)[0]
    embeddings_s = embeddings[select_idx]
    return embeddings_s

```

```python
import numpy as np
import pandas as pd
from scipy.spatial.distance import cdist
from scipy.stats import ks_2samp
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(0)
nobias_embed = get_embedding(embeddings_array_nb, annotation_nb, "AT1")
sim1_embed = get_embedding(embeddings_array_nbp1, annotation_nb, "AT1")


# Calculate pairwise Euclidean distances
dist_nobias = cdist(nobias_embed, nobias_embed, 'euclidean')
dist_sim1 = cdist(sim1_embed, sim1_embed, 'euclidean')
# Flatten the distance matrices
dist_nobias_flat = dist_nobias[np.triu_indices_from(dist_nobias, k=1)]
dist_sim1_flat = dist_sim1[np.triu_indices_from(dist_sim1, k=1)]
# Perform Kolmogorov-Smirnov test
ks_stat, p_val = ks_2samp(dist_nobias_flat, dist_sim1_flat)
print(f'KS test: KS-statistic = {ks_stat}, p-value = {p_val}')
# PCA for visualization
pca = PCA(n_components=2)
nobias_pca = pca.fit_transform(nobias_embed)
sim1_pca = pca.fit_transform(sim1_embed)
# Plot PCA results
plt.figure(figsize=(10, 6))
plt.scatter(nobias_pca[:, 0], nobias_pca[:, 1], alpha=0.7, label='nobias')
plt.scatter(sim1_pca[:, 0], sim1_pca[:, 1], alpha=0.7, label='sim1')
plt.title('PCA of Embeddings')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.show()
```
