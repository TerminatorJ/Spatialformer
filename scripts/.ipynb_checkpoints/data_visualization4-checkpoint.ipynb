```python
import pandas as pd
import pygwalker as pyg
#loading the data
df = pd.read_csv("/home/sxr280/Spatialformer/data/all48samples_info.csv",index_col = 0)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = df[df["count"] < 100000]

# Normalize 'mean_gene_count' values for color mapping
df['normalized_mean_gene_count'] = (df['mean_gene_count'] - df['mean_gene_count'].min()) / (
    df['mean_gene_count'].max() - df['mean_gene_count'].min()
)

# Define your desired cutoff values
cutoff_count = 100
cutoff_mean_gene_count = 2
cutoff_unique_gene_count = 10

df['normalized_mean_gene_count'] = df['normalized_mean_gene_count'].where(
    df['mean_gene_count'] >= cutoff_mean_gene_count, 0
)

df['normalized_mean_gene_count'] = df['normalized_mean_gene_count'].where(
    df['count'] >= cutoff_count, 0
)

df['normalized_mean_gene_count'] = df['normalized_mean_gene_count'].where(
    df['unique_gene_count'] >= cutoff_unique_gene_count, 0
)
# Define color mapping (gray for values below cutoff, blue gradient for values above)
colors = ['blue' if value == 0 else "blue" for value in df['normalized_mean_gene_count']]
# Create a figure and axes
fig, ax = plt.subplots(figsize=(10, 10))

# Scatter plot of 'count' vs. 'unique_gene_count'
ax.scatter(df['count'], df['unique_gene_count'], s=20, c=colors, alpha=0.7)

# Set labels and title
ax.set_xlabel('count', fontsize=12)
ax.set_ylabel('unique_gene_count', fontsize=12)
ax.set_title('Count vs. Unique Gene Count', fontsize=14)

# Add horizontal and vertical lines for cutoffs
# ax.axvline(x=cutoff_count, color='red', linestyle='--')
# ax.axhline(y=cutoff_gene_count, color='red', linestyle='--')
# Add a colorbar to show the mapping of colors to 'mean_gene_count'
# sm = plt.cm.ScalarMappable(cmap=plt.cm.Blues, norm=plt.Normalize(vmin=0, vmax=1))
# sm.set_array([])
# cbar = fig.colorbar(sm, label='Mean Gene Count', shrink=0.5)

plt.savefig("/home/sxr280/Spatialformer/Figures/data_filtering_scatter_plot.png", dpi=300)
# Show the plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define parameters
max_num_epochs = 1000  # You can adjust this to your needs

# Generate epoch values
epochs = np.arange(0, max_num_epochs + 1)  # From 0 to max_num_epochs

# Calculate lr_factor for each epoch
lr_factors = 0.5 * (1 + np.cos(np.pi * epochs / max_num_epochs))

# Plotting the lr_factor
plt.figure(figsize=(10, 5))
plt.plot(epochs, lr_factors, label='Learning Rate Factor', color='blue')
plt.xlabel('Epochs')
plt.ylabel('Learning Rate Factor')
plt.title('Cosine Learning Rate Factor over Epochs')
plt.xlim(0, max_num_epochs)
plt.ylim(0, 1)  # lr_factor ranges from 0 to 1
plt.grid()
plt.legend()
plt.show()
```
