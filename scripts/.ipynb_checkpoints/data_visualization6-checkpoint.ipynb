Attach the annotation to the adata


```python
import os
import json
def getann(sample_name):
    datadir = "/scratch/project_465001027/Spatialformer/data"
#     ann_files = [file for file in os.listdir(datadir) if "annotations" in file]
#     niche_files = [file for file in os.listdir(datadir) if "niches" in file]
    ctfile = os.path.join(datadir, f"{sample_name}_annotations.json")
    nichefile = os.path.join(datadir, f"{sample_name}_niches.json")
    with open(ctfile, "r") as f1:
        ct_dict = json.load(f1)
    with open(nichefile, "r") as f2:
        nichefile = json.load(f2)
    return ct_dict, nichefile
```

```python
sample_name = "THD0008"

#loading the annotation information
ct_dict, nichefile = getann(sample_name)
#get annotation in the sdata object
sdata.table.obs['Annotations'] = sdata.table.obs['cell_id'].apply(lambda x: ct_dict.get(x))
sdata.table.obs['Niches'] = sdata.table.obs['cell_id'].apply(lambda x: nichefile.get(x))
sdata.tables["table"] = sdata.table[sdata.table.obs['Annotations'].notna(),:]
```

plotting the network image for the gene-gene occurrence in a specific sample

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# Assume genepair_adata is your AnnData object with columns named ("gene1","gene2")
data_matrix = genepair_adata.X.toarray()  # Convert matrix to numpy array

# Initialize graph
G = nx.Graph()

# Extract gene pair from column names and compute node sizes
node_sizes = {}

for gene_pair in genepair_adata.var_names:
    # Split gene pair
    gene1, gene2 = eval(gene_pair)  # or use ast.literal_eval(gene_pair) for safer evaluation
    
    # Count occurrences
    occurrence_count = data_matrix[:, genepair_adata.var_names.get_loc(gene_pair)].sum()

    # Update/add nodes
    if gene1 not in node_sizes:
        node_sizes[gene1] = 0
    if gene2 not in node_sizes:
        node_sizes[gene2] = 0
        
    node_sizes[gene1] += occurrence_count
    node_sizes[gene2] += occurrence_count

    # Add edge between gene1 and gene2 if the pair exists in any cell
    if occurrence_count > 0:
        G.add_edge(gene1, gene2, weight=occurrence_count)

import csv

# Create CSV for edges
with open('/scratch/project_465001027/Spatialformer/data/gene_interactions_edges.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['source', 'target', 'weight'])
    
    for u, v, data in G.edges(data=True):
        writer.writerow([u, v, data['weight']])

# Create CSV for nodes
with open('/scratch/project_465001027/Spatialformer/data/gene_interactions_nodes.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['gene', 'size'])

    for gene, size in node_sizes.items():
        writer.writerow([gene, size])

print("CSV files created: gene_interactions_edges.csv and gene_interactions_nodes.csv")

```

```python
def filterann(adata, ann):
    # Create a copy of the data to avoid modifying the original data directly
    adata_copy = adata.copy()
    adata.obs["Annotations"] = adata.obs["Annotations"].astype('category')
    # Update the 'Annotations' column, setting it to 'others' where it does not match the provided annotation
    adata_copy.obs['Annotations'] = adata_copy.obs['Annotations'].apply(lambda x: ann if x == ann else 'others')
    adata_copy.uns.pop('Annotations_colors', None)  # Remove existing color settings
    # Define a color palette with two colors, for the annotation of interest and 'others'
#     colors = sns.color_palette("Set1", 2)  # You can choose a different palette or customize colors
#     adata_copy.uns['Annotations_colors'] = colors
    #get the color map
    colors = adataflt.uns["Annotations_colors"]
    #get the ann color
    ann_color = colors[list(adataflt.obs["Annotations"].cat.categories).index(ann)]
    new_color = [ann_color, "#FBF9FF"]
    adata_copy.uns['Annotations_colors'] = new_color
    return adata_copy


    
```
