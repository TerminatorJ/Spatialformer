```python
def load_model(model_ckp_path, device):
    get_file_path = lambda path, filename: os.path.join("/scratch/project_465001027/Spatialformer", path, filename)
    config_path = get_file_path("config", "_config_train_large_pair.json")
    with open(config_path, 'r') as json_file:
        config = json.load(json_file)
    # model_ckp_path = "/scratch/project_465001027/Spatialformer/output/checkpoints/step=0100000-train_total_loss=-2.2727-val_total_loss=0.0000.ckpt"
    model = manual_train_fm(config = config)
    ckp = torch.load(model_ckp_path, map_location=torch.device(device))
    params = ckp["state_dict"]
    model.load_state_dict(params)
    model.eval()
    model.to(device)
    return model
```

```python
model_ckp_path = "/scratch/project_465001027/Spatialformer/output/checkpoints/step=0100000-train_total_loss=-2.2727-val_total_loss=0.0000.ckpt"
model = load_model(model_ckp_path, "cuda")
#loading the sample dataset
combined_dataset = load_from_disk("/scratch/project_465001027/Spatialformer/cache/xenium_pandavid_dataset4")  
combined_dataset_all = concatenate_datasets([combined_dataset["train"], combined_dataset["test"], combined_dataset["validation"]])
index_path = "/scratch/project_465001027/Spatialformer/data/sample_cell_index.pkl"
sample_cell_index = get_index(combined_dataset, save_file = index_path)
```

```python
df = pd.read_excel("/scratch/project_465001027/Spatialformer/data/Supplementary_Table1.xlsx", sheet_name='Human_data')
dv_df = pd.read_excel("/scratch/project_465001027/Spatialformer/data/Supplementary_Table1.xlsx", sheet_name='David_dataset', header = 1)
def assign(sample):
    tissues = []
    conditions = []
    for sample_name in sample["Sample_Names"]:

        if sample_name in list(dv_df["Sample"]):
            tissue = "Lung"
            condition = dv_df[dv_df["Sample"] == sample_name]["Status"].values[0]
        else:
            tissue = df[df["Str_ID"] == sample_name]["Tissues"].values[0]
            condition = df[df["Str_ID"] == sample_name]["Condition2"].values[0]
        tissues.append(tissue)
        conditions.append(condition)
    return {"Tissues": tissues, "Conditions": conditions}
#getting the tissues and condition information
combined_dataset_tmp = combined_dataset_all.map(assign, batched = True, batch_size = 300, num_proc = 256)
# Saving the dataset with only only two columns
meta_columns = combined_dataset_tmp.select_columns(["Tissues","Conditions"])
meta_columns.save_to_disk("/scratch/project_465001027/Spatialformer/data/meta_columns", num_proc = 256)
```

Generating the embeddings

```python
%%time
all_embeds1 = []
all_embeds2 = []
all_conditions = []
all_tissues = []
all_anns = []
counter = 0
#select 1% of the sample to do visualization
selected_num = int(combined_dataset_all.shape[0] * 0.02)
# selected_num = 3
with torch.no_grad(): 
    for i, batch in tqdm(enumerate(dataloader), total = int(selected_num/batch_size)):
        counter += batch_size
        tissues = batch["Tissues"]
        conditions = batch["Conditions"]
        anns = batch["Annotations"]
        embeddings, ref_pair_prob = model.get_embeddings(batch, [-1], True, False) #normal prob
        #getting the conditions and tissue types
        embeddings1 = torch.mean(embeddings[0][:,5:,:], dim = 1).squeeze() #b x dim
#         embeddings = embeddings[0][:,0,:]
        embeddings1 = embeddings1.detach().cpu().numpy()#The embeddings should 
    
        all_embeds1.append(embeddings1)
        
        embeddings2 = embeddings[0][:,0,:].detach().cpu().numpy()
        all_embeds2.append(embeddings2)
        
        all_tissues.append(tissues)
        all_anns.append(anns)
        all_conditions.append(conditions)
        if counter >= selected_num:
            break
```

```python

import time
import matplotlib.pyplot as plt
import umap
import umap.plot
import matplotlib.pyplot as plt
import networkx as nx
import community.community_louvain as community_louvain
 

normal_label_dict = {}
simulation_label_dict = {}
for i,ann in enumerate(np.unique(annotation)[:10]):
    #filter the clusters
    print(f"doing {ann}")
    select_idx = np.where(np.array(annotation_nb) == ann)[0]
    embeddings_s = embeddings_array_nb[select_idx]
    mapper = umap.UMAP(n_neighbors = 10, min_dist = 0.5).fit(embeddings_s)
    knn_graph = umap.umap_.fuzzy_simplicial_set(
    X=embeddings_s,
    n_neighbors=100,
    random_state=np.random.RandomState(),
    metric='euclidean')[0]

    # Convert knn_graph to a networkx graph
    G = nx.Graph(knn_graph)

    # Step 4: Apply Louvain clustering
    louvain_partition = community_louvain.best_partition(G)

    # Extract the cluster labels
    cluster_labels = np.array(list(louvain_partition.values()))

    # Step 5: Plot the UMAP embedding with Louvain clusters
    fig1, ax1 = plt.subplots()
    print("normal")
    normal_label_dict[ann] = len(np.unique(cluster_labels))
    fig1 = umap.plot.points(mapper, labels=cluster_labels, color_key_cmap='Set1')
    plt.show()
#####################################################################################
    select_idx = np.where(np.array(annotation_nbp) == ann)[0]
    embeddings_s = embeddings_array_nbp[select_idx]
    mapper = umap.UMAP(n_neighbors = 10, min_dist = 0.5).fit(embeddings_s)
    knn_graph = umap.umap_.fuzzy_simplicial_set(
    X=embeddings_s,
    n_neighbors=100,
    random_state=np.random.RandomState(),
    metric='euclidean')[0]

    # Convert knn_graph to a networkx graph
    G = nx.Graph(knn_graph)

    # Step 4: Apply Louvain clustering
    louvain_partition = community_louvain.best_partition(G)

    # Extract the cluster labels
    cluster_labels = np.array(list(louvain_partition.values()))

    # Step 5: Plot the UMAP embedding with Louvain clusters
    fig2, ax2 = plt.subplots()
    print("simulation")
    simulation_label_dict[ann] = len(np.unique(cluster_labels))
    fig2 = umap.plot.points(mapper, labels=cluster_labels, color_key_cmap='Set1')
    plt.show()
    plt.close()
    # time.sleep(3)


```

```python
def get_embedding(embeddings, annotation, cluster):
    select_idx = np.where(np.array(annotation) == cluster)[0]
    embeddings_s = embeddings[select_idx]
    return embeddings_s

```

```python
import numpy as np
import pandas as pd
from scipy.spatial.distance import cdist
from scipy.stats import ks_2samp
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(0)
nobias_embed = get_embedding(embeddings_array_nb, annotation_nb, "AT1")
sim1_embed = get_embedding(embeddings_array_nbp1, annotation_nb, "AT1")


# Calculate pairwise Euclidean distances
dist_nobias = cdist(nobias_embed, nobias_embed, 'euclidean')
dist_sim1 = cdist(sim1_embed, sim1_embed, 'euclidean')
# Flatten the distance matrices
dist_nobias_flat = dist_nobias[np.triu_indices_from(dist_nobias, k=1)]
dist_sim1_flat = dist_sim1[np.triu_indices_from(dist_sim1, k=1)]
# Perform Kolmogorov-Smirnov test
ks_stat, p_val = ks_2samp(dist_nobias_flat, dist_sim1_flat)
print(f'KS test: KS-statistic = {ks_stat}, p-value = {p_val}')
# PCA for visualization
pca = PCA(n_components=2)
nobias_pca = pca.fit_transform(nobias_embed)
sim1_pca = pca.fit_transform(sim1_embed)
# Plot PCA results
plt.figure(figsize=(10, 6))
plt.scatter(nobias_pca[:, 0], nobias_pca[:, 1], alpha=0.7, label='nobias')
plt.scatter(sim1_pca[:, 0], sim1_pca[:, 1], alpha=0.7, label='sim1')
plt.title('PCA of Embeddings')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.show()
```
